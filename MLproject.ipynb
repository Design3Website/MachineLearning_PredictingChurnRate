```python
from sklearn.ensemble import IsolationForest
# Detection des outliers dans le data set

iso = IsolationForest(contamination=0.05)  # 5% de données aberrantes
yhat = iso.fit_predict(df[df.columns])
outliers_iso = df[yhat == -1]
outliers_percentage_iso = (len(outliers_iso) / len(df)) * 100
print(f'Nombre des outliers detectés: {len(outliers_iso)}')
print(f'Pourcentage des outliers detectés: {outliers_percentage_iso:.2f}%')
outliers_iso.head()
```

```python
from sklearn.neighbors import LocalOutlierFactor

# Detection des outliers avec LOF

lof = LocalOutlierFactor(n_neighbors=20)
yhat = lof.fit_predict(df[df.columns])
outliers_lof = df[yhat == -1]
outliers_percentage = (len(outliers_lof) / len(df)) * 100
print(f'Nombre des outliers detectés: {len(outliers_lof)}')
print(f'Pourcentage des outliers detectés: {outliers_percentage:.2f}%')
outliers_lof.head()
```

  <ul style="margin-left:20px;">
    <strong> <li> Tests statistiques </li></strong>

  </ul>
  <p> <strong> Test de Welch  </strong> sur les variables continues indiquent si la moyenne d'une variable varie significativement d'un groupe à l'autre. L'hypothèse (H0) est que les moyennes d'une variable sont les mêmes dans les classes. On peut ici partout rejeter (H0) autrement dit supposer que les variables sont toutes discriminantes pour les classes.</p>
</div>

```python
from scipy.stats import ttest_ind

quantitative_cols = ['Tenure', 'Balance', 'CreditScore', 'Age' , 'NumOfProducts']

ttest_results = []

for col in quantitative_cols:
    group_0 = df[df['Churn'] == 0][col]
    group_1 = df[df['Churn'] == 1][col]
    
    t_stat, p = ttest_ind(group_0, group_1, equal_var=False)  # Welch test
    
    ttest_results.append({
        'Variable': col,
        'Test': 'Welch t-test',
        't_stat': t_stat,
        'p_value': p,
        'Mean_churn_0': group_0.mean(),
        'Mean_churn_1': group_1.mean()
    })

ttest_df = pd.DataFrame(ttest_results)
print(ttest_df.sort_values('p_value'))
```

<div style="font-family: Georgia, 'Times New Roman', serif; font-size:15px; line-height:1.6;">

```python
from scipy.stats import chi2_contingency
import itertools
# Préparer une liste des variables qualitatives

cols =['Geography' , 'Gender' , 'IsActiveMember' , 'HasCrCard', 'Churn']
# Créer une liste pour stocker les résultats
results = []

# Boucle sur toutes les paires de variables
for var1, var2 in itertools.combinations(cols, 2):
    # Créer le tableau croisé
    table = pd.crosstab(df[var1], df[var2])
    
    # Test du chi-deux
    chi2, p, dof, expected = chi2_contingency(table)
    
    # Ajouter les résultats
    results.append({
        'Variable 1': var1,
        'Variable 2': var2,
        'Chi2': chi2,
        'p-value': p,
        'dof': dof
    })

# Convertir en DataFrame pour synthèse
chi2_summary = pd.DataFrame(results)
print(chi2_summary)

```

<div style="font-family: Georgia, 'Times New Roman', serif; font-size:15px; line-height:1.6;">

  
  <p> <strong> Interprétation des p-values  </strong> 
  On peut supposer que les variables catégoriques explicatives sont 2 à 2 indépendantes (petites p_values). Par contre elles sont toutes dépendantes à Churn à l'exception de HasCrCard. Si on se focalise HasCrCard seule, cette variable explicative n'influence pas le Churn à priori mais son effet combiné à d'autres variables pourrait être sensible.

<div style="font-family: Georgia, 'Times New Roman', serif; font-size:15px; line-height:1.6;">

  <h3 style="margin-top:15px;">
    Définition des métriques de performance pour la sélection des modèles
  </h3>

  <p>
    Généralement, les métriques utilisées pour évaluer un modèle de machine learning sont les suivants :
  </p>

  <ul style="margin-left:20px;">
    <li>
      <strong>Accuracy</strong> : mesure la proportion globale de prédictions correctes :
      <br>
      <em>(TP + TN) / (TP + TN + FP + FN)</em>
    </li>
    <li>
      <strong>Precision</strong> : mesure la proportion de prédictions positives correctes :
      <br>
      <em>TP / (TP + FP)</em>
    </li>
    <li>
      <strong>Recall (Sensitivity)</strong> : mesure la capacité du modèle à identifier correctement
      les observations positives :
      <br>
      <em>TP / (TP + FN)</em>
    </li>
    <li>
      <strong>F1-score</strong> : moyenne harmonique entre la précision et le rappel,
      particulièrement pertinente dans le cas de <strong>classes déséquilibrées</strong> :
      <br>
      <em>2 × (Precision × Recall) / (Precision + Recall)</em>
    </li>
    <li>
      <strong>Courbe ROC</strong> : représente le <em>True Positive Rate (Recall)</em>
      en fonction du <em>False Positive Rate</em> pour l’ensemble des seuils de décision.
    </li>
    <li>
      <strong>AUC (Area Under the Curve)</strong> : mesure la capacité globale du modèle

```python

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score

# --- Fonction pour un petit modèle simple ---
def train_simple_logit(X, y, scale_numeric=False):
    # Split train/test stratifié
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, stratify=y, random_state=42
    )
    
    # Scaling si nécessaire
    if scale_numeric:
        scaler = StandardScaler()
        X_train = pd.DataFrame(scaler.fit_transform(X_train), columns=X_train.columns, index=X_train.index)
        X_test = pd.DataFrame(scaler.transform(X_test), columns=X_test.columns, index=X_test.index)
    
    # Entraîner la régression logistique simple
    model = LogisticRegression(class_weight='balanced', solver='liblinear', max_iter=1000)
    model.fit(X_train, y_train)
    
    # Prédictions et F1 score
    y_pred = model.predict(X_test)
    rc = recall_score(y_test , y_pred)
    f1 = f1_score(y_test, y_pred)
    
    # Affichage des résultats
    print("\n--- Modèle avec {} ---".format(", ".join(X.columns)))
    print("Coefficient :", pd.Series(model.coef_[0], index=X.columns))
    print("Recall :", rec)
    print("F1 score :", f1)
    
    return model, X_train, X_test, y_train, y_test

# --- Préparer la variable cible ---
y = df_encoded['Churn']

# --- 1️⃣ Age seulement ---
X_age = df_encoded[[col for col in df_encoded.columns if col.startswith('Age_class_')]]
train_simple_logit(X_age, y, scale_numeric=False)

# --- 2️⃣ NumOfProducts seulement ---
X_numprod = df_encoded[[col for col in df_encoded.columns if col.startswith('NumProducts_class_')]]
train_simple_logit(X_numprod, y, scale_numeric=False)

# --- 3️⃣ CreditScore seulement ---
X_credit = df_encoded[['CreditScore']]
train_simple_logit(X_credit, y, scale_numeric=True)

```
